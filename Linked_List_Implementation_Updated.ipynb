```python
class Node:
  def __init__(self,data):
    self.data=data
    self.next=None

  def traverse(first_node):
    cur=first_node
    while cur:
      print(cur.data,", ")
      cur=cur.next
    print("null")
```

```python
node1 = Node(7)
node2 = Node(11)
node3 = Node(3)
node4 = Node(4)
node5 = Node(9)

node1.next = node2
node2.next = node3
node3.next = node4
node4.next = node5

```

```python
# Traverse a list
```

```python

Node.traverse(node1)
```

```python
# Find the lowest value
```

```python
def lowest_value(first_node):
    cur=first_node
    lowest=first_node.data

    if not first_node:
      print("Empty list")



    while cur:
      if lowest>cur.data:
        lowest=cur.data
      cur=cur.next


    print("Lowest value: ",lowest)

```

```python
lowest_value(node1)
```

```python
# Delete a specific node
```

```python
def delete_node(first_node,data):
  cur=first_node

  if not first_node:
    print("Empty list")

  if first_node.data == data:
      return first_node.next



  while cur.next and cur.next.data!=data:
    cur=cur.next

  if cur.next==None:
    print("Data is not present")
    return

  cur.next=cur.next.next

  return cur.data










```

```python
delete_node(node1,11)

```

```python
Node.traverse(node1)
```

```python
# Inserting a node
```

```python
def insert(start_node,data,pos):

  if pos==1:
    data.next=start_node
    return data

  cur=start_node

  for _ in range(pos-2):
    if cur.next is None:
      break
    cur=cur.next

  data.next=cur.next
  cur.next=data

  return start_node
```

```python
node1 = Node(7)
node2 = Node(3)
node3 = Node(2)
node4 = Node(9)

node1.next = node2
node2.next = node3
node3.next = node4
```

```python
newNode = Node(97)
node1 = insert(node1, newNode, 2)
```

```python
Node.traverse(node1)
```
