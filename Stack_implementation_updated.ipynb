```
class Stack:

  def __init__(self):
    self.stack=[]

  def push(self,item):
    self.stack.append(item)

  def pop(self):
    if self.isEmpty():
      return "Empty stack"
    else:
      return self.stack.pop()

  def isEmpty(self):
    return len(self.stack)==0

  def peek(self):
    if self.isEmpty():
      return "Empty stack"
    return self.stack[-1]

  def size(self):
    return len(self.stack)
```

```
myStack=Stack()

myStack.push(1)
myStack.push(2)
myStack.push(3)
myStack.push(4)
myStack.push(5)

```

```
print("Stack: ", myStack.stack)
```

```
print("Pop: ", myStack.pop())
print("Stack after Pop: ", myStack.stack)
print("Peek: ", myStack.peek())
print("isEmpty: ", myStack.isEmpty())
print("Size: ", myStack.size())
```
